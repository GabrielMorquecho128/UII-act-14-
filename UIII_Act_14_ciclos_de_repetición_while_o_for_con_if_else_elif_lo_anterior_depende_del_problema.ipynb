## Calcula la suma de la serie armónica: 1 + 1/2 + 1/3 + ... + 1/N.

```python
# ======================================================
# 🧮 CBTis 128
# 📚 Elabora Proyectos de Programacion Logica
# 🤖 Especialidad: Inteligencia Artificial
# 👨‍🏫 Docente: Ing. Eliseo Nava
# 👨‍🎓 Alumno: Marco Morquecho
# 👥 Grupo: 2H
# 📘 Submodulo: Programacion
# 🔢 Problema 43:
# Calcular la suma de la serie armonica
# 1 + 1/2 + 1/3 + ... + 1/N
# ======================================================

print("==========================================")
print("   SUMA DE LA SERIE ARMONICA")
print("==========================================")

nombre = input("Ingresa tu nombre: ")

N = int(input("Ingresa el valor de N: "))

suma = 0

i = 1

while i <= N:

    suma = suma + (1 / i)

    print("1/", i, "=", 1 / i)

    i += 1

# ======================================================

print("\n==========================================")
print(" RESULTADO FINAL ")
print("==========================================")

print("Alumno:", nombre)

print("Valor de N:", N)

print("Resultado de la suma:", suma)

# ======================================================

if suma >= 2:

    print("La suma armonica es grande")

elif suma >= 1:

    print("La suma armonica es media")

else:

    print("La suma armonica es pequeña")

print("\nPrograma finalizado")
```
